## 如何在 Kaggle 上使用（How to Use on Kaggle）

1. 打开上方任意一个实验链接；
2. 点击右上角 **Copy & Edit** 复制到自己的账户下；
3. 将 **Accelerator（加速器）** 设置为 **None (CPU)**；
4. 仅实验五需要：在 **Notebook Settings → Internet** 中开启 Internet（用于下载约 11 MB 的 MNIST 数据集）；
5. 点击 **Run All** 一键运行全部代码。

每个实验都是完全自包含的——无需上传数据集、无需 API 密钥、无需任何本地文件。
